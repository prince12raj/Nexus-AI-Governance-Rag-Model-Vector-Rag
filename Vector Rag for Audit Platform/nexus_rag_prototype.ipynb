```python
def create_chunks(text, chunk_size=150, overlap=30):
    words = str(text).split()
    chunks = []

    start = 0

    while start < len(words):
        end = start + chunk_size
        chunk = " ".join(words[start:end])

        if chunk.strip():
            chunks.append(chunk)

        if end >= len(words):
            break

        start = end - overlap

    return chunks
```

```python
chunk_records = []

for _, row in rag_df.iterrows():

    chunks = create_chunks(row["text"])

    for chunk_number, chunk in enumerate(chunks, start=1):

        chunk_records.append({
            "framework": row["framework"],
            "clause_id": row["clause_id"],
            "chunk_id": chunk_number,
            "input_text": row["input_text"],
            "text": chunk
        })

chunks_df = pd.DataFrame(chunk_records)

print("Original records:", len(rag_df))
print("Total chunks:", len(chunks_df))

display(chunks_df.head())
```

```python
import faiss
import pandas as pd
from sentence_transformers import SentenceTransformer

# Load embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Load FAISS index
loaded_index = faiss.read_index(str(faiss_file))

# Load metadata
loaded_metadata = pd.read_csv(metadata_file)

print("FAISS vectors:", loaded_index.ntotal)
print("Metadata rows:", len(loaded_metadata))
```

```python
def retrieve(query, k=5):
    
    # Convert query into embedding
    query_embedding = model.encode(
        [query],
        convert_to_numpy=True
    ).astype("float32")
    
    # Search FAISS
    distances, indices = loaded_index.search(
        query_embedding,
        k
    )
    
    results = []
    
    for rank, idx in enumerate(indices[0]):
        
        if idx == -1:
            continue
        
        row = loaded_metadata.iloc[idx]
        
        results.append({
            "rank": rank + 1,
            "distance": float(distances[0][rank]),
            "framework": row["framework"],
            "clause_id": row["clause_id"],
            "chunk_id": row["chunk_id"],
            "text": row["text"]
        })
    
    return pd.DataFrame(results)
```

```python
query = "Can a company keep customer personal data indefinitely?"

results = retrieve(query, k=5)

display(results)
```

```python
for _, row in results.iterrows():
    
    print("=" * 100)
    print("Rank:", row["rank"])
    print("Distance:", row["distance"])
    print("Framework:", row["framework"])
    print("Clause ID:", row["clause_id"])
    print("Chunk ID:", row["chunk_id"])
    print("\nRetrieved Text:")
    print(row["text"])
    print()
```

```python
def build_context(query, k=5):
    
    results = retrieve(query, k)
    
    context_parts = []
    
    for _, row in results.iterrows():
        
        context_parts.append(
            f"""Framework: {row['framework']}
Clause ID: {row['clause_id']}
Chunk ID: {row['chunk_id']}
Text: {row['text']}"""
        )
    
    return "\n\n".join(context_parts)
```

```python
query = "How should an organization handle personal data retention?"

context = build_context(query, k=5)

print(context)
```

```python
compliance_rules = {
    "data_retention": [
        "retain personal data indefinitely",
        "delete old customer records",
        "retention period",
        "storage limitation",
        "delete personal data",
        "erase personal data"
    ],
    
    "data_collection": [
        "collect personal data",
        "necessary personal data",
        "purpose of collection",
        "data minimization"
    ],
    
    "data_security": [
        "protect personal data",
        "security measures",
        "unauthorized access",
        "encryption"
    ]
}

print("Compliance rules loaded:", len(compliance_rules))
```

```python
def detect_compliance_area(policy):
    policy_lower = policy.lower()
    
    matched_areas = []
    
    for area, keywords in compliance_rules.items():
        for keyword in keywords:
            if keyword in policy_lower:
                matched_areas.append(area)
                break
    
    return matched_areas


areas = detect_compliance_area(policy)

print("Detected compliance areas:")
for area in areas:
    print("-", area)
```

```python
def rule_based_audit(policy, k=5):
    results = retrieve(policy, k=k)
    
    areas = detect_compliance_area(policy)
    
    if len(results) == 0:
        verdict = "INSUFFICIENT EVIDENCE"
        risk = "UNKNOWN"
    else:
        if "data_retention" in areas:
            verdict = "REVIEW REQUIRED"
            risk = "HIGH"
        else:
            verdict = "REVIEW REQUIRED"
            risk = "MEDIUM"
    
    evidence = []
    
    for _, row in results.iterrows():
        evidence.append({
            "framework": row["framework"],
            "record_id": int(row["clause_id"]),
            "chunk_id": int(row["chunk_id"]),
            "distance": round(float(row["distance"]), 4),
            "text": row["text"]
        })
    
    return {
        "policy": policy.strip(),
        "compliance_areas": areas,
        "verdict": verdict,
        "risk": risk,
        "evidence": evidence
    }
```

```python
audit_report = rule_based_audit(policy, k=5)

print("Verdict:", audit_report["verdict"])
print("Risk:", audit_report["risk"])
print("Areas:", audit_report["compliance_areas"])
```

```python
for item in audit_report["evidence"]:
    print("\nFramework:", item["framework"])
    print("Record ID:", item["record_id"])
    print("Chunk ID:", item["chunk_id"])
    print("Distance:", item["distance"])
    print("Evidence:", item["text"])
```

```python
def calculate_compliance_score(audit_report):
    score = 100
    
    if audit_report["risk"] == "HIGH":
        score -= 40
    elif audit_report["risk"] == "MEDIUM":
        score -= 20
    
    if audit_report["verdict"] == "INSUFFICIENT EVIDENCE":
        score = 0
    
    return max(0, score)


score = calculate_compliance_score(audit_report)

print("Compliance Score:", score, "/ 100")
```

```python
def generate_recommendations(audit_report):
    recommendations = []
    
    if "data_retention" in audit_report["compliance_areas"]:
        recommendations.append(
            "Define a clear retention period for customer personal data."
        )
        recommendations.append(
            "Automatically delete or anonymize personal data when it is no longer required."
        )
    
    if "data_collection" in audit_report["compliance_areas"]:
        recommendations.append(
            "Collect only personal data that is necessary for the stated purpose."
        )
    
    if "data_security" in audit_report["compliance_areas"]:
        recommendations.append(
            "Implement appropriate security measures to protect personal data."
        )
    
    if not recommendations:
        recommendations.append(
            "Review the policy against the available regulatory evidence."
        )
    
    return recommendations


recommendations = generate_recommendations(audit_report)

print("Recommendations:")
for i, recommendation in enumerate(recommendations, 1):
    print(f"{i}. {recommendation}")
```
